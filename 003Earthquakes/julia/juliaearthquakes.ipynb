```julia
using Pkg
Pkg.add("CSV")
Pkg.add("HTTP")
Pkg.add("DataFrames")
using HTTP, CSV, DataFrames
using Base.Filesystem: isfile
println("Imports ✅")
```

```julia
function create_folders(folders::Vector{String})
    for folder in folders
        if !isdir(folder)
            mkdir(folder)
        end
    end
    println("Folders ✅")
end;
```

```julia
function download_data(url, filepath)
    if isfile(filepath)
        println("Data loaded from existing file 📁")
        return
    end
    try
        response = HTTP.get(url)
        open(filepath, "w") do f
            write(f, response.body)
        end
        println("Data ✅")
    catch error
        println("Error ❌: $error")
    end
end;
```

```julia
url = "https://earthquake.usgs.gov/fdsnws/event/1/query?format=csv&starttime=2000-01-01&endtime=2000-12-31&minmagnitude=4.5"
filepath = joinpath("data", "query.csv")
folders= ["data", "results"]
create_folders(folders)
download_data(url, filepath)
data = CSV.File(filepath) |> DataFrame;
```
